# The four basic plots

## Line plot

```python
a = [1, 2, 3, 4]
b = [1, 4, 3, 4]
plt.plot(a, b)
```

## Dashed line style

```python
plt.plot(a, b, ls="--")
```

## colored line

```python
plt.plot(a, b, color="purple")
```

## Swap x and y

```python
plt.plot(b, a)
```

## Multiple lines at once

```python
df = pd.DataFrame()
df['mass'] = [7, 10, 13, 17]
df['ext'] = [3, 3, 3, 5]
df['width'] = [1, 8, 11, 7]
df
```

```python
# plots each column
df.plot()
```

## Bar Chart

```python
a = ['A','B','C']
b = [4, 1, 2]
plt.bar(a, b)
```

## Horizontal bar Chart

```python
a = ['A','B','C']
b = [4, 1, 2]
plt.barh(a, b)
```

## Scatter plot

```python
a = [1, 2, 3, 4]
b = [1, 4, 3, 4]
plt.scatter(a, b)
```

### colored points

```python
plt.scatter(a, b, color='red')
```

### Square markers

```python
plt.scatter(a, b, marker='s')
```

### Stars markers

```python
plt.scatter(a, b, marker='*')
```

```python
#triangle markers
plt.scatter(a, b, marker='^',color='green')
```

### Semi-transparent markers

```python
plt.scatter(a, b, alpha=0.5)
```

### Diamond markers

```python
plt.scatter(a, b, marker='D')
```

### fully styled

```python
plt.scatter(a, b, marker='o',color='purple',alpha=0.6)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('fully styled scatter plot')
```

# Plotting DataFrames

```python
import pandas as pd
df = pd.DataFrame()
df['layer'] = ["crust","mantle","outer core","inner core"]
df['thickness'] = [40,2900,2200,1230]
df
```

```python
x = df['layer']
y = df['thickness']
plt.bar(x, y)
```

# Labels and titles

```python
plt.bar( df['layer'], df['thickness'] )
plt.ylabel('Thickness (km)')
plt.xlabel('Layer')
plt.title('Layers of the Earth')
```

### Adjust vertical padding of title

```python
plt.bar( df['layer'], df['thickness'] )
plt.ylabel('Thickness (km)')
plt.xlabel('Layer')
plt.title('Layers of the Earth',pad=30)
```

```python
print(df.columns.tolist())
```

### Smaller title

```python
plt.bar( df['layer'], df['thickness'] )
plt.ylabel('Thickness (km)')
plt.xlabel('Layer')
plt.title('Layers of the Earth',fontsize=7)
```

### size change font labels

```python
plt.bar( df['layer'], df['thickness'] )
plt.ylabel('Thickness (km)',fontsize=15)
plt.xlabel('Layer',fontsize=20)
plt.title('Layers of the Earth',fontsize=7)
```
